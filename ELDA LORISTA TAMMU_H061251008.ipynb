Tampilkan proyeksi stereonet bola bawah (*beachball diagram*) serta tentukan parameter sesar (*strike*, *dip*, *rake*).

#### ***Detail Komponen {R,T,C,I,K,O,Q,V}***

Berperan sebagai *Observational Seismologist* dan Pakar Seismotektonik yang berpengalaman dalam analisis penentuan mekanisme sumber gempa (*source mechanism*) dan dekomposisi tensor momen.

2. Menghitung komponen matriks Tensor Momen Seismik $M_{ij}$ 3x3 simetris.
3. Menghitung Momen Seismik Skalar ($M_0$) dan Magnitudo Momen ($M_w$) berdasarkan persamaan Kanamori.
4. Menentukan orientasi bidang sesar (*nodal planes*): *strike*, *dip*, dan *rake*.
5. Plot diagram *beachball* menggunakan proyeksi bola bawah (*lower-hemisphere projection*).

##### **C — Context**
Penentuan mekanisme sumber gempa bumi sangat krusial untuk memahami pola deformasi regional dan mengidentifikasi orientasi patahan aktif pemicu gempa bumi.

##### **I — Input**
- Tabel stasiun seismik berisi: Nama Stasiun, Azimuth (derajat), Take-off Angle (derajat), dan Polaritas gelombang P (+1 / -1).
- Nilai rigiditas batuan medium ($\mu = 30 \text{ GPa}$).
- Buatkan dataset sintetis untuk 12 stasiun jaringan seismik jika data riil belum tersedia.

##### **K — Constraints**
1. Gunakan Python dengan pustaka `obspy.imaging.beachball` (atau `matplotlib` / `scipy`).
2. Proyeksi harus menggunakan *equal-area projection* (Schmidt net).
3. Nilai $M_w$ dihitung eksplisit dengan rumus $M_w = \frac{2}{3} \log_{10}(M_0) - 10.7$ (dengan $M_0$ dalam dyne-cm).

##### **O — Output**
1. **Tabel Parameter Sesar**: Orientasi Plane 1 dan Plane 2 (*Strike/Dip/Rake*).
2. **Gambar Diagram Beachball**: Plot proyeksi stereonet dengan simbol stasiun *compressional* dan *dilatational*.
3. **Perhitungan Matematis**: Langkah detail pembentukan tensor momen $M_{ij}$ dan energi seismik.
4. **Kode Python**: Skrip lengkap dan modular.

##### **Q — Quality Criteria**
- Diagram *beachball* menyajikan secara jelas pemisahan kuadran kompresi dan dilatasi.
- Nilai $M_w$ konsisten secara matematis dengan skalar momen $M_0$.

##### **V — Verification**

### ***PROMPT 2 : TEKTONOFISIKA (RELEMBATAN SESAR / COULOMB STRESS TRANSFER)***
**Judul**: Pemodelan Perubahan Tegangan Coulomb (*Coulomb Failure Stress Transfer*) dan Analisis Reaktivasi Sesar Aktif.

**Instruksi**: Bertindaklah sebagai Pakar Tektonofisika. Hitung perubahan tegangan geser dan normal pada bidang pemicu akibat gempabumi utama (*mainshock*), lalu petakan distribusi perubahan Tegangan Coulomb ($\Delta CFS$) pada grid 2D menggunakan Python untuk mengidentifikasi area pemicuan gempa susulan (*aftershock trigger zone*).

#### ***Detail Komponen {R,T,C,I,K,O,Q,V}***

##### **R — Role**
Berperan sebagai *Geodynamicist* & Pakar Tektonofisika Komputasi yang berfokus pada mekanika batuan, dinamika kerak bumi, dan transfer tegangan tektonik.

##### **T — Task**
1. Menghitung medan deformasi dan elastisitas dari model dislokasi Okada 1985.
2. Menghitung perubahan tegangan geser ($\Delta \tau$) dan tegangan normal ($\Delta \sigma_n$) pada bidang sasaran (*receiver fault*).
3. Menghitung perubahan tegangan patah Coulomb: $\Delta CFS = \Delta \tau + \mu' \Delta \sigma_n$.
4. Memetakan zona pemicuan positif ($\Delta CFS > 0$) dan zona bayangan tegangan ($\Delta CFS < 0$).

##### **C — Context**
Setiap kali gempa bumi besar terjadi, retakan pada patahan melepaskan tegangan di satu area namun memindahkan akumulasi tegangan ke area di sekitarnya. Pemodelan $\Delta CFS$ membantu memprediksi lokasi gempa susulan atau pemicuan patahan tetangga.

##### **I — Input**
- Parameter Sesar Sumber (*Source Fault*): Panjang ($L$), Lebar ($W$), *Slip* ($u$), *Strike*, *Dip*, *Rake*, Kedalaman ($z$).
- Koefisien gesek efektif batuan ($\mu' = 0.4$).
- Modulus geser batuan ($G = 30 \text{ GPa}$) dan Poisson's ratio ($\nu = 0.25$).

##### **K — Constraints**
1. Gunakan Python dengan pustaka `numpy`, `scipy`, dan `matplotlib`.
2. Nilai $\Delta CFS$ harus disajikan dalam satuan bar atau kiloPascals ($\text{kPa}$).
3. Grid kalkulasi bidang horizontal berukuran minimal $50 \times 50$ titik pada kedalaman hiposenter.

##### **O — Output**
1. **Peta Kontur 2D $\Delta CFS$**: Visualisasi dengan skala warna terpisah (Merah untuk pemicuan / positif, Biru untuk bayangan tegangan / negatif).
2. **Analisis Sensitivitas**: Pengaruh perubahan nilai koefisien gesek $\mu'$ terhadap zona pemicuan.
3. **Kode Python**: Kode simulasi transfer tegangan lengkap.

##### **Q — Quality Criteria**
- Peta $\Delta CFS$ menampilkan pola simetri empat kuadran (*quadrupole pattern*) yang khas untuk sesar geser.
- Unit dan konversi tegangan dari Pa ke bar / kPa tepat.

##### **V — Verification**
Verifikasi bahwa area tepat di ujung lobus patahan mengalami peningkatan $\Delta CFS$ positif (zona merah), sedangkan area lateral mengalami penurunan tegangan.

### ***PROMPT 3 : GEOMATH / MATEMATIKA GEOFISIKA (INVERSI NON-LINIER MATRIKS LEVENBERG-MARQUARDT)***
**Judul**: Solusi Inversi Non-Linier Menggunakan Algoritma Levenberg-Marquardt / Damped Least-Squares untuk Pemodelan Parameter Geofisika.

**Instruksi**: Bertindaklah sebagai Dosen Matematika Geofisika. Tulis skrip Python untuk menyelesaikan persoalan inversi non-linier menggunakan matriks Jacobian dan regularisasi Damped Least-Squares (Levenberg-Marquardt). Terapkan algoritma ini untuk mengestimasi parameter fisik medium dari kurva respons observasi yang terdistorsi oleh *noise*.

#### ***Detail Komponen {R,T,C,I,K,O,Q,V}***

##### **R — Role**
Berperan sebagai *Geophysical Mathematician* dan Pakar Teori Inversi Geofisika yang menguasai pemodelan matematika, kalkulus multivariabel, dan aljabar linier terapan.

##### **T — Task**
1. Memformulasi fungsi pemodelan maju (*forward modeling function*) $d = f(m)$.
2. Menghitung matriks Jacobian ($J$) secara analitik atau diferensial terhingga (*finite-difference*).
3. Menyusun iterasi pembaruan parameter: $\Delta m = (J^T J + \lambda I)^{-1} J^T (d_{obs} - f(m))$.
4. Mengimplementasikan penghentian otomatis (*stopping criteria*) berbasis penurunan RMS error.
5. Menampilkan grafik konvergensi error terhadap jumlah iterasi.

##### **C — Context**
Hampir seluruh masalah geofisika bersifat *ill-posed* dan non-linier. Pendekatan matematika geofisika modern membutuhkan algoritma inversi ter-damped untuk memastikan keterbatasan solusi dan stabilitas komputasi.

##### **I — Input**
- Vektor data observasi $d_{obs}$ berjumlah $N$ sampel yang diberi *noise* Gaussian 5%.
- Model awal tebakan (*initial guess*) parameter $m_0$.
- Parameter dambatan awal (damping factor $\lambda_0 = 0.1$).

##### **K — Constraints**
1. Gunakan pustaka murni Python: `numpy` dan `matplotlib` (tanpa menggunakan pustaka *blackbox optimization* seperti `scipy.optimize.curve_fit` agar proses matriks transparan).
2. Iterasi dibatasi maksimum 50 langkah atau hingga RMS $< 10^{-4}$.

##### **O — Output**
1. **Penurunan Formula**: Penjelasan rinci turunan matriks Jacobian $J$ dan persamaan Levenberg-Marquardt.
2. **Grafik Fitting**: Kurva Data Observasi vs Model Awal vs Hasil Inversi Akhir.
3. **Grafik Konvergensi**: Plot Iterasi vs Root Mean Square Error (RMSE).
4. **Kode Python**: Kode aljabar matriks bersih terstruktur.

##### **Q — Quality Criteria**
- Algoritma stabil dan tidak mengalami *divergence* akibat pemulihan nilai $\lambda$.
- Parameter estimasi mendekati nilai parameter sebenarnya (*true model*) dalam batas toleransi *noise*.

##### **V — Verification**
Pastikan ukuran matriks Jacobian sesuai ($N \times M$, dengan $N$ = jumlah data, $M$ = jumlah parameter model) dan periksa perkalian matriks $J^T J$ menghasilkan matriks persegi $M \times M$.

### ***PROMPT 4 : METODE SEISMIK REFRAKSI (METODE PLUS-MINUS / HAGEDOORN UNTUK LAPISAN MIRING)***
**Judul**: Pemodelan Kedalaman Bedrock Miring dan Tomografi Waktu Tiba Seismik Refraksi Menggunakan Metode *Plus-Minus* / *Reciprocal*.

**Instruksi**: Bertindaklah sebagai Ahli Geofisika Eksplorasi. Lakukan pengolahan data seismik refraksi untuk kasus lapisan bawah permukaan miring menggunakan *Plus-Minus Method*. Hitung kecepatan gelombang $P$ lapisan pertama ($V_1$), lapisan kedua ($V_2$), serta kedalaman di bawah tiap geofon ($z_x$). Sajikan penampang lintasan 2D beserta skrip Python.

#### ***Detail Komponen {R,T,C,I,K,O,Q,V}***

##### **R — Role**
Berperan sebagai *Near-Surface Exploration Geophysicist* yang menguasai akuisisi, pengolahan sinyal, dan interpretasi metode seismik refraksi dangkal.

##### **T — Task**
1. Mengolah data waktu tiba gelombang refraksi (*first arrival time*) dari skema penembakan *Forward*, *Reverse*, dan *Over-shot*.
2. Menghitung waktu *Plus* ($T_+$) dan waktu *Minus* ($T_-$) untuk setiap lokasi geofon.
3. Menghitung nilai kecepatan refraktor $V_2$ dari kemiringan kurva $T_-$ vs jarak ($x$).
4. Menentukan kedalaman batas perlapisan di bawah setiap geofon ($z_x$) berdasarkan waktu $T_+$.
5. Merekonsiliasi model penampang kedalaman 2D (*geological cross-section*).

##### **C — Context**
Metode konvensional Intercept Time mengasumsikan lapisan rata/horisontal. Namun, pada kondisi topografi dan kedalaman batuan dasar (*bedrock*) yang tidak teratur, metode *Plus-Minus* Hagedoorn diperlukan untuk memperoleh profil kedalaman di setiap titik geofon secara presisi.

##### **I — Input**
- Posisi geofon $x$ (0 hingga 100 m, interval 5 m).
- Waktu tiba tembakan Forward $t_{A}(x)$, Reverse $t_{B}(x)$, dan waktu timbal balik total $T_{AB}$.
- Buat data sintetis realistis untuk lapisan tanah ($V_1 = 600 \text{ m/s}$) di atas batuan dasar miring ($V_2 = 2200 \text{ m/s}$) jika data nyata belum ada.

##### **K — Constraints**
1. Persamaan $T_+$ dan $T_-$ harus didefinisikan ketat sesuai teori Hagedoorn: $T_+ = t_{Ax} + t_{Bx} - T_{AB}$ dan $T_- = t_{Ax} - t_{Bx} + T_{AB}$.
2. Gunakan Python dengan pustaka `numpy`, `pandas`, dan `matplotlib`.
3. Satuan jarak dalam meter ($m$), waktu dalam milidetik ($ms$), dan kecepatan dalam $m/s$.

##### **O — Output**
1. **Tabel Pengolahan**: Posisi Geofon, $t_A$, $t_B$, $T_+$, $T_-$, dan Kedalaman $z_x$.
2. **Grafik Kurva Waktu-Jarak**: Plot $T-X$ lengkap dengan titik *crossover*.
3. **Penampang Kedalaman 2D**: Profil topografi permukaan dan kontur batas *bedrock* bawah permukaan.
4. **Kode Python**: Kode komputasi lengkap terstruktur.

##### **Q — Quality Criteria**
- Kecepatan $V_2$ dihitung via regresi linier slope $T_-$ vs $x$ dengan kesesuaian $R^2 > 0.98$.
- Penampang 2D menampilkan ketebalan lapisan secara kontinu di sepanjang lintasan.

##### **V — Verification**
Verifikasi bahwa $T_{AB}$ konstan untuk tembakan maju dan balik, serta periksa bahwa nilai $T_+$ selalu bernilai positif.

### ***PROMPT 5 : GEOSTATISTIKA (VARIOGRAM FITTING & ORDINARY KRIGING SPASIAL 2D)***
**Judul**: Pemodelan Variogram Eksperimen dan Interpolasi *Ordinary Kriging* 2D untuk Pemetaan Properti Geofisika Spasial.

**Instruksi**: Bertindaklah sebagai Pakar Geostatistika. Buatlah skrip Python untuk menghitung variogram eksperimental spasial dari data pengukuran lapangan ($x, y, z$). Fit model variogram teoritis (Spherical / Exponential), lalu terapkan algoritma *Ordinary Kriging* untuk melakukan estimasi inter-titik beserta peta ketidakpastian (*kriging variance*).

#### ***Detail Komponen {R,T,C,I,K,O,Q,V}***

##### **R — Role**
Berperan sebagai *Spatial Geostatistician* dan Pakar Pemodelan Karakterisasi Cadangan yang berpengalaman dalam analisis korelasi spasial dan estimasi bobot Kriging.

##### **T — Task**
1. Menghitung nilai semi-variogram eksperimental $\gamma(h) = \frac{1}{2N(h)} \sum (Z(x) - Z(x+h))^2$.
2. Melakukan pemodelan variogram teoritis (*fitting*) untuk mengekstrak parameter *Nugget* ($C_0$), *Sill* ($C_0 + C$), dan *Range* ($a$).
3. Menyusun dan menyelesaikan sistem persamaan matriks Kriging: $K \cdot w = k$.
4. Menginterpolasi nilai pada grid 2D yang belum terukur beserta memetakan varians estimasi ($\sigma_k^2$).

##### **C — Context**
Data geofisika lapangan selalu tersebar secara spasial pada titik-titik sampel terbatas. Interpolasi sederhana seperti IDW (*Inverse Distance Weighting*) sering mengabaikan korelasi spasial dan tidak memberikan nilai ketidakpastian, sehingga metode *Ordinary Kriging* menjadi standar industri geofisika.

##### **I — Input**
- Data spasial titik acak berisi kolom: Koordinat $X$ (meter), Koordinat $Y$ (meter), dan Nilai Properti $Z$ (misal: Nilai Kecepatan $V_p$ atau Kedalaman Bedrock).
- Generasikan 30 titik sampel acak berserta nilainya jika data belum dilampirkan.

##### **K — Constraints**
1. Gunakan pustaka Python: `PyKrige` (atau `scipy.spatial` / `numpy` murni) dan `matplotlib`.
2. Matriks Kriging wajib menyertakan pengali Lagrange ($\mu$) untuk menjamin kondisi tak-bias (*unbiasedness constraint* $\sum w_i = 1$).

##### **O — Output**
1. **Plot Variogram**: Grafik lag distance vs semivariance beserta kurva model teoritis.
2. **Peta Estimasi Kriging 2D**: Peta kontur warna hasil interpolasi properti geofisika.
3. **Peta Ketidakpastian (Variance Map)**: Peta kontur $\sigma_k^2$ yang menunjukkan area dengan tingkat kepercayaan rendah.
4. **Kode Python**: Kode geostatistika terstruktur lengkap.

##### **Q — Quality Criteria**
- Varians Kriging berharga nol pada titik lokasi sampel asli.
- Model variogram teoritis memilih nilai *Nugget*, *Sill*, dan *Range* yang meminimalkan eror fitting.

##### **V — Verification**
Verifikasi bahwa jumlah bobot Kriging pada setiap titik estimasi tepat bernilai 1.0 ($\sum w_i = 1$) dan nilai varians Kriging meningkat seiring menjauhnya jarak dari lokasi stasiun sampel.
