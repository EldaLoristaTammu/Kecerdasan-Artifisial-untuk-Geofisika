## 1. Audit Kelengkapan dan Konsistensi Data Stasiun CSV

### Tujuan
Menyusun pemeriksaan awal untuk mengetahui apakah data stasiun dapat digunakan sebagai masukan analisis. Pemeriksaan berfokus pada susunan kolom, isi data, pencatatan waktu, dan kemungkinan duplikasi, tanpa mengubah file mentah.

### Landasan Singkat
Kualitas hasil analisis sangat dipengaruhi oleh kualitas data yang digunakan. Kolom yang tidak tersedia, waktu yang gagal dibaca, atau rekaman berulang dapat memengaruhi pengolahan berikutnya. Oleh karena itu, proses audit sebaiknya menghasilkan daftar temuan yang dapat ditelusuri, bukan langsung mengubah data yang dianggap bermasalah.

### Masukan dan Batasan
- `INPUT_CSV`: alamat file CSV yang akan ditinjau.
- `TIMESTAMP_FIELD`: nama kolom tanggal/waktu.
- `EXPECTED_FIELDS`: kolom yang diharapkan tersedia.
- `FIELD_UNITS`: satuan variabel, jika relevan.
- File sumber harus tetap utuh.
- Data kosong, tidak terbaca, atau berulang diberi penanda dalam laporan.
- Kesimpulan dibatasi pada hal yang benar-benar dapat diperiksa dari tabel.

### Prompt dan Prosedur
**Prompt penggunaan ulang:**

> Lakukan audit pada file `[INPUT_CSV]` sebelum data dipakai untuk analisis. Periksa keberadaan kolom yang dibutuhkan, tipe data, kemampuan membaca timestamp, rentang waktu, sel kosong, serta catatan waktu yang muncul lebih dari sekali. Sajikan hasilnya sebagai laporan pemeriksaan dan tunjukkan bagian yang perlu ditinjau manual. Jangan mengedit file sumber dan jangan menyebut data valid secara keseluruhan jika hanya sebagian pemeriksaan yang berhasil.

**Prosedur kerja:**
1. Muat CSV dalam mode baca dan catat lokasi file yang diperiksa.
2. Tampilkan ukuran tabel serta nama kolom.
3. Bandingkan kolom yang tersedia dengan `EXPECTED_FIELDS`.
4. Uji konversi kolom waktu dan catat nilai yang gagal dikonversi.
5. Hitung jumlah nilai kosong pada tiap kolom.
6. Identifikasi timestamp berulang dan periksa apakah baris lainnya juga identik.
7. Susun laporan dengan status lulus, peringatan, atau gagal untuk setiap pemeriksaan.

### Rencana Implementasi
Gunakan `pandas` untuk membaca tabel, memeriksa kolom, menghitung nilai kosong, serta meninjau timestamp. Jika konversi waktu dilakukan, simpan hasilnya sebagai objek sementara untuk pemeriksaan; jangan menulis kembali file mentah. Nama kolom dan format waktu harus disesuaikan dengan data yang sebenarnya.

### Keluaran yang Direncanakan
- `station_audit.md`: laporan hasil pemeriksaan.
- Tabel ringkasan jumlah baris, kolom, dan nilai kosong.
- Daftar kolom yang tidak ditemukan dan waktu yang tidak dapat dibaca.
- Ringkasan timestamp berulang serta catatan masalah yang perlu ditindaklanjuti.

### Pemeriksaan Hasil
Cocokkan jumlah baris pada laporan dengan tabel yang dibaca dan pastikan file sumber tidak berubah. Bila ada kolom waktu yang tidak sesuai, laporan harus menyebutkan hal itu secara jelas. Gunakan status **Not yet executed** jika kode baru direncanakan dan belum dijalankan.

## 2. Validasi Struktur dan Atribut Data GeoJSON

### Tujuan
Membuat prosedur untuk meninjau kualitas file GeoJSON dan menghasilkan inventaris fitur agar isi data spasial dapat diketahui sebelum digunakan dalam pemetaan atau analisis lanjutan.

### Landasan Singkat
GeoJSON menyimpan fitur geografis beserta geometri dan atributnya. Pemeriksaan struktur dapat memastikan apakah file dapat dibaca dan apakah objek penting tersedia. Namun, file yang terbaca belum tentu memiliki koordinat yang benar, geometri yang valid, atau informasi sistem koordinat yang lengkap.

### Masukan dan Batasan
- `SOURCE_GEOJSON`: lokasi data GeoJSON.
- Lokasi penyimpanan hasil validasi dan inventaris.
- Atribut penting yang perlu dipertahankan.
- Informasi CRS atau sumber metadata jika tersedia.
- Jangan menghapus fitur hanya karena atributnya kosong tanpa aturan yang ditetapkan.
- Jangan memperbaiki koordinat berdasarkan dugaan.
- Semua perubahan yang dilakukan harus dapat dijelaskan dan dibandingkan dengan sumber.

### Prompt dan Prosedur
**Prompt penggunaan ulang:**

> Tinjau `[SOURCE_GEOJSON]` untuk mengetahui struktur dan kelengkapan data spasial. Hitung jumlah fitur, inventarisasi tipe geometri, periksa atribut yang tersedia, identifikasi geometri kosong atau koordinat yang tidak dapat dibaca, dan catat informasi CRS jika tercantum. Buat laporan yang membedakan kesalahan terkonfirmasi dari hal yang memerlukan pengecekan tambahan. Pertahankan file asli dan jangan menyatakan hasil validasi lengkap sebelum seluruh pemeriksaan yang direncanakan selesai.

**Prosedur kerja:**
1. Baca file dan pastikan sintaks JSON dapat diurai.
2. Periksa tipe objek utama dan keberadaan daftar fitur.
3. Hitung jumlah fitur serta kelompokkan berdasarkan tipe geometri.
4. Buat daftar nama atribut dan frekuensi atribut yang tidak terisi.
5. Periksa koordinat yang hilang atau memiliki struktur tidak sesuai.
6. Catat informasi CRS dari metadata atau dokumentasi sumber; jangan menebaknya.
7. Buat inventaris fitur dan laporan validasi.
8. Jika ada proses normalisasi, bandingkan hasilnya dengan data awal dan jelaskan setiap perubahan.

### Rencana Implementasi
Pustaka `json` dapat digunakan untuk memeriksa struktur dasar, sedangkan `geopandas` atau pustaka geospasial lain dapat dipakai jika tersedia untuk membaca dan memeriksa geometri. Ketergantungan perangkat lunak dan versi yang digunakan perlu dicatat pada saat implementasi.

### Keluaran yang Direncanakan
- `geojson_validation.md`: ringkasan validasi.
- `feature_inventory.csv`: daftar fitur dan informasi geometri/atribut.
- GeoJSON hasil penataan, hanya jika memang dilakukan dan lolos pemeriksaan.
- Catatan fitur yang bermasalah atau memerlukan tinjauan manual.

### Pemeriksaan Hasil
Pastikan file keluaran dapat dibaca kembali, jumlah fitur dapat dijelaskan, dan atribut penting tidak hilang. Bila jumlah fitur berubah setelah penataan, dokumentasikan alasan perubahan. Pemeriksaan struktur tidak boleh dianggap sebagai bukti tunggal bahwa posisi geografis seluruh fitur sudah benar.

## 3. Pemodelan Numerik Difusi Panas Satu Dimensi

### Tujuan
Merancang simulasi sederhana untuk mengamati penyebaran panas sepanjang suatu material dari kondisi suhu awal hingga waktu yang ditentukan. Model menggunakan parameter termal, kondisi batas, serta pembagian ruang dan waktu.

### Landasan Singkat
Persamaan difusi panas satu dimensi dinyatakan sebagai:

\[
\frac{\partial T}{\partial t} = \alpha \frac{\partial^2 T}{\partial x^2}
\]

Keterangan:
- `T`: suhu.
- `t`: waktu.
- `x`: posisi.
- `α`: difusivitas termal.

Untuk skema eksplisit beda hingga yang umum, bilangan kestabilan dapat dihitung dengan:

\[
r = \frac{\alpha \Delta t}{(\Delta x)^2}
\]

Pada skema eksplisit standar satu dimensi, nilai `r` umumnya perlu memenuhi `r ≤ 0,5`. Syarat ini bergantung pada skema numerik yang diterapkan, sehingga harus diverifikasi sebelum simulasi dilakukan.

### Masukan dan Batasan
- `T0`: distribusi suhu pada waktu awal.
- `THERMAL_DIFFUSIVITY`: nilai difusivitas dan satuannya.
- `LEFT_BOUNDARY` dan `RIGHT_BOUNDARY`: kondisi pada ujung domain.
- `DX`: jarak antar-node ruang.
- `DT`: interval waktu numerik.
- `SIMULATION_DURATION`: waktu total simulasi.
- Parameter harus memakai sistem satuan yang konsisten.
- Hasil numerik tidak boleh disebut sebagai hasil eksperimen atau kondisi lapangan tanpa data pendukung.

### Prompt dan Prosedur
**Prompt penggunaan ulang:**

> Susun model difusi panas satu dimensi dengan suhu awal `[T0]`, difusivitas `[THERMAL_DIFFUSIVITY]`, kondisi batas pada kedua ujung, jarak grid `[DX]`, interval waktu `[DT]`, dan durasi `[SIMULATION_DURATION]`. Periksa konsistensi satuan dan kestabilan metode sebelum menghitung. Tampilkan perubahan profil suhu pada beberapa waktu serta jelaskan asumsi model. Pisahkan hasil perhitungan, parameter masukan, dan interpretasi.

**Prosedur kerja:**
1. Periksa nilai parameter, satuan, dan bentuk profil suhu awal.
2. Tentukan rentang posisi serta titik-titik grid.
3. Susun langkah waktu sesuai durasi yang ditentukan.
4. Terapkan kondisi awal dan kondisi batas pada grid.
5. Hitung bilangan kestabilan sebelum iterasi dimulai.
6. Lakukan pembaruan suhu berdasarkan skema numerik yang dipilih.
7. Simpan profil pada waktu tertentu dan buat grafik perbandingan.
8. Bandingkan perilaku hasil dengan kasus sederhana yang dapat diperkirakan.
9. Catat parameter, peringatan, dan status pelaksanaan simulasi.

### Rencana Implementasi
Gunakan `numpy` untuk perhitungan array dan `matplotlib` untuk membuat grafik profil suhu. Implementasi harus menjelaskan formula pembaruan suhu dan penanganan batas. Jika parameter belum diberikan, gunakan placeholder dan jangan membuat klaim hasil numerik tertentu.

### Keluaran yang Direncanakan
- Grafik distribusi suhu terhadap posisi untuk beberapa waktu.
- Tabel nilai suhu pada titik grid yang dipilih.
- Ringkasan parameter dan bilangan kestabilan.
- Catatan asumsi, peringatan, serta status eksekusi.

### Pemeriksaan Hasil
Pastikan suhu awal dan kondisi batas diterapkan dengan benar, dimensi array sesuai, dan simulasi tidak berjalan ketika syarat kestabilan skema yang dipilih tidak terpenuhi. Perubahan `DX` dan `DT` dapat dipakai untuk memeriksa sensitivitas hasil, tetapi tidak otomatis membuktikan konvergensi.

## 4. Estimasi Lokasi Episenter Berdasarkan Waktu Tiba Gelombang P dan S

### Tujuan
Menyusun workflow untuk mengolah catatan waktu tiba gelombang P dan S dari beberapa stasiun seismik, memperkirakan jarak sumber gempa terhadap stasiun, dan memvisualisasikan perkiraan lokasi episenter. Recipe ini merupakan topik baru yang berbeda dari pemeriksaan CSV, GeoJSON, difusi panas, maupun pemetaan angin.

### Landasan Singkat
Gelombang P umumnya tiba lebih dahulu daripada gelombang S. Selisih waktu tiba dapat ditulis:

\[
\Delta t = T_S - T_P
\]

Selisih tersebut berkaitan dengan jarak antara sumber gempa dan stasiun, tetapi konversinya memerlukan asumsi atau model kecepatan gelombang. Jika jarak perkiraan dari beberapa stasiun tersedia, lingkaran berpusat pada setiap stasiun dapat digambarkan untuk membantu menilai area perpotongan. Hasil ini merupakan estimasi awal episenter, bukan penentuan kedalaman hiposenter secara otomatis.

### Masukan dan Batasan
- `STATION_TABLE`: tabel nama stasiun dan koordinatnya.
- `P_ARRIVAL`: waktu tiba gelombang P di tiap stasiun.
- `S_ARRIVAL`: waktu tiba gelombang S di tiap stasiun.
- `VELOCITY_MODEL`: asumsi/model kecepatan gelombang yang digunakan untuk mengubah selisih waktu menjadi jarak.
- Sistem koordinat dan zona waktu harus konsisten.
- Data waktu yang hilang atau urutan waktu yang tidak masuk akal harus ditandai.
- Jangan menentukan lokasi pasti hanya dari satu stasiun.
- Ketidakpastian waktu picking dan model kecepatan perlu dicatat.

### Prompt dan Prosedur
**Prompt penggunaan ulang:**

> Analisis tabel waktu tiba gelombang P dan S dari `[STATION_TABLE]`. Validasi koordinat stasiun, format waktu, nilai kosong, dan konsistensi bahwa waktu tiba S tidak lebih awal daripada waktu tiba P. Hitung `Δt = Ts - Tp` untuk data yang memenuhi syarat. Gunakan `[VELOCITY_MODEL]` yang dinyatakan secara eksplisit untuk memperkirakan jarak, lalu buat visualisasi lingkaran jarak dari masing-masing stasiun. Tampilkan area perpotongan sebagai perkiraan episenter dan jelaskan keterbatasan hasil. Jangan mengarang koordinat sumber jika data tidak cukup.

**Prosedur kerja:**
1. Baca data stasiun, koordinat, dan waktu tiba gelombang.
2. Pastikan semua stasiun memiliki identitas dan koordinat yang dapat digunakan.
3. Validasi format waktu serta tandai catatan P/S yang tidak lengkap.
4. Hitung selisih waktu `Ts - Tp` untuk tiap stasiun yang valid.
5. Gunakan model kecepatan yang dipilih untuk mengubah selisih waktu menjadi jarak perkiraan.
6. Ubah koordinat ke sistem lokal yang sesuai untuk visualisasi jika diperlukan, sambil mencatat metode konversinya.
7. Gambar posisi stasiun dan lingkaran dengan radius sesuai jarak perkiraan.
8. Tinjau area yang paling konsisten dengan lingkaran dari beberapa stasiun.
9. Laporkan perkiraan lokasi, stasiun yang digunakan, asumsi model, serta sumber ketidakpastian.

### Rencana Implementasi
`pandas` dapat dipakai untuk membaca dan memvalidasi tabel, `numpy` untuk perhitungan, dan `matplotlib` untuk menggambar stasiun serta lingkaran jarak. Untuk wilayah yang luas, perhitungan koordinat dan jarak sebaiknya menggunakan metode geodesik atau proyeksi yang tepat, bukan menganggap derajat lintang dan bujur setara dengan jarak meter. Model kecepatan harus dipilih berdasarkan informasi yang tersedia dan dinyatakan di laporan.

### Keluaran yang Direncanakan
1. Tabel QA/QC waktu tiba P dan S.
2. Tabel selisih waktu dan jarak perkiraan per stasiun.
3. Plot lokasi stasiun dan lingkaran jarak.
4. Ringkasan area perkiraan episenter beserta stasiun yang digunakan.
5. Catatan asumsi, keterbatasan, dan status pelaksanaan.

### Pemeriksaan dan Kriteria Keberhasilan
- Koordinat setiap stasiun dapat ditelusuri ke sumber.
- Waktu P dan S menggunakan format serta referensi waktu yang sama.
- Selisih waktu dihitung dengan benar dan nilai tidak valid ditandai.
- Konversi selisih waktu menjadi jarak mencantumkan model kecepatan yang digunakan.
- Plot memiliki legenda, skala atau konteks koordinat yang jelas.
- Lokasi dilaporkan sebagai estimasi jika ketidakpastian belum dianalisis secara lengkap.

### Keterbatasan
Akurasi estimasi dipengaruhi oleh sebaran geografis stasiun, ketelitian picking, struktur kecepatan bawah permukaan, dan metode konversi jarak. Lingkaran yang tidak bertemu pada satu titik dapat menunjukkan ketidakpastian data atau keterbatasan model. Penentuan hiposenter tiga dimensi umumnya memerlukan informasi tambahan, seperti kedalaman dan pemodelan waktu tempuh yang lebih lengkap.
